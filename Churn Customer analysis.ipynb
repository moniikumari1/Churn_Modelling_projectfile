# Data Analysis

```python
copy=churn_customer.copy()
```

```python
def categorize_creditscore(CreditScore):
    if CreditScore<=597:
        return"Poor"
    elif CreditScore<=669:
        return"Fair"
    elif CreditScore<=739:
        return"Good"
    elif CreditScore<=799:
        return"Very Good"
    else:
        return"Ecellent"
```

```python
copy["categorize_of_creditscore"]=copy["CreditScore"].apply(categorize_creditscore)
```

```python
copy["HasCrCard"]=copy["HasCrCard"].replace({1:"Yes",0:"No"})
```

```python
copy["IsActiveMember"]=copy["IsActiveMember"].replace({1:"IsActive",0:"NoActive"})
```

```python
copy
```

```python
# Total Customer
copy["CustomerId"].count()
```

```python
# Total Churn Customer
copy[copy["Exited"]==1]["CustomerId"].count()
```

```python
# Churn Rate
round(copy[copy["Exited"]==1]["CustomerId"].count()/copy["CustomerId"].count()*100,2)
```

```python
# Average Balance of Churn Customer
copy[copy["Exited"]==1]["Balance"].mean().round(2)
```

```python
# Estimate Salary of Churn Customer
copy[copy["Exited"]==1]["EstimatedSalary"].mean().round()
```

```python
# Geogrphy by churn Rate
copy.groupby("Geography")["Exited"].mean().mul(100).round(2)
```

```python
# Churn Rate by Gendar
copy.groupby("Gender")["Exited"].mean().mul(100).round(2)
```

```python
# Churn Rate by Age category
copy.groupby("Age_group")["Exited"].mean().mul(100).round(2)
```

```python
# Churn rate by Cateory of Credit Score
copy.groupby("categorize_of_creditscore")["Exited"].mean().mul(100).round(2)
```

```python
#  Churn rate  by Tenure
copy.groupby("Tenure")["Exited"].mean().mul(100).round(2)
```

```python
# Churn Rate by No. of product
copy.groupby("NumOfProducts")["Exited"].mean().mul(100).round(2)
```

```python
# Churn Rate by Has CreditCard 
copy.groupby("HasCrCard")["Exited"].mean().mul(100).round(2)
```

```python
# Churn Rate by Active member
copy.groupby("IsActiveMember")["Exited"].mean().mul(100).round(2)
```

# Visualization

```python
customer=copy["Exited"].value_counts()
customer.index=["Non churn","churn"]
print(customer)
```

```python
plt.figure(figsize=(4,4))
plt.bar(customer.index,customer.values)
plt.title("Customer by Churn and Non Churn Customer")
plt.xlabel("Customer")
plt.ylabel("Number of Customer")
```

```python
plt.figure(figsize=(5,3))
geo=copy.groupby("Geography")["Exited"].mean().mul(100).round(2).sort_values(ascending=False).reset_index()
plt.barh(geo["Geography"],geo["Exited"],color="green")
plt.title("Churn Rate by Geography")
plt.xlabel("Churn Rate")
plt.ylabel("Geographhy")
plt.show()
```

```python
plt.figure(figsize=(4,4))
gen=copy.groupby("Gender")["Exited"].mean().mul(100).round(2).reset_index()
sns.barplot(x="Gender",y="Exited",data=gen,palette="pastel")
plt.title("Churn Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Churn Rate")
plt.show()
```

```python
cat=copy.groupby("categorize_of_creditscore")["Exited"].mean().mul(100).round(2).reset_index()
plt.plot(cat["categorize_of_creditscore"],cat["Exited"],marker='o')
plt.title("Churn Rate by Categories of Credit Card")
plt.xlabel("Categories of Credit Card")
plt.ylabel("Churn Rate")
plt.show()
```

```python
age=copy.groupby("Age_group")["Exited"].mean().mul(100).round(2).sort_values(ascending=False).reset_index()
plt.barh(age["Age_group"],age["Exited"],color="purple")
plt.gca().invert_yaxis()
plt.title("Churn Rate by Age Group")
plt.xlabel("Churn Rate")
plt.ylabel(" Age Group")
plt.show()
```

```python
plt.figure(figsize=(10,3))
ten=copy.groupby("Tenure")["Exited"].mean().mul(100).round(2).reset_index()
sns.lineplot(x="Tenure",y="Exited",data=ten,marker='o',color='orange')
plt.title("Churn Rate by Tenure")
plt.show()
```

```python
num=copy.groupby("NumOfProducts")["Exited"].mean().mul(100).round(2).reset_index()
sns.barplot(x="NumOfProducts",y="Exited",data=num,color="skyblue")
plt.title("Churn Rate by Number of product")
plt.show()
```

```python
plt.figure(figsize=(4,3))
cre=copy.groupby("HasCrCard")["Exited"].mean().mul(100).round(2).reset_index()
plt.barh(cre["HasCrCard"],cre["Exited"],color="blue")
plt.title("Churn Rate by Has Credit card")
plt.xlabel("Churn Rate")
plt.ylabel("Has Credit card")
plt.show()
```

```python
plt.figure(figsize=(4,3))
mem=copy.groupby("IsActiveMember")["Exited"].mean().mul(100).round(2).reset_index()
plt.barh(mem["IsActiveMember"],mem["Exited"],color="brown")
plt.title("Churn Rate by Is Active Member")
plt.xlabel("Churn Rate")
plt.ylabel("Is Active Member")
plt.show()
```
